```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = 1.0
N = 100000
dt = t / N
M = 10000
tiempo = np.linspace(0, t, N+1)

integrales = np.zeros(M)

for i in range(M):
    dW = np.random.normal(0, np.sqrt(dt), N)
    W = np.zeros(N+1)
    W[1:] = np.cumsum(dW)

    integral = np.sum(W[:-1]**2) * dt
    integrales[i] = integral

media_simulada = np.mean(integrales)
varianza_simulada = np.var(integrales)

print(f"Media teórica: 0.5, Media simulada: {media_simulada:.4f}")
print(f"Varianza teórica: {1/3:.4f}, Varianza simulada: {varianza_simulada:.4f}")

plt.hist(integrales, bins=50, density=True, alpha=0.6, color='g')
plt.title('Distribución de \(\int_0^1 W_s^2 ds\)')
plt.xlabel('Valor de la integral')
plt.ylabel('Densidad')
plt.show()
```
