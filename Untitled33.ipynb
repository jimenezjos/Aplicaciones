### Ecuación Diferencial Parcial (EDP) de Black-Scholes

La ecuación general para el precio de una opción $V(S,t)$ es:

$
 \frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + (r - q)S \frac{\partial V}{\partial S} - rV = 0
$

Sustituyendo los datos del problema $\sigma = 0.29$, $r = 0.04$, $q = 0$:

$
 \frac{\partial V}{\partial t} + 0.04205 S^2 \frac{\partial^2 V}{\partial S^2} + 0.04 S \frac{\partial V}{\partial S} - 0.04 V = 0
 $\
Con la condición terminal para una opción Call con $E = 60$ y $T = 0.3$:

$
 V(S, 0.3) = \max(S - 60, 0)
$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
S, E, T, r, sigma, q = sp.symbols('S E T r sigma q', real=True, positive=True)

datos = {
    S: 58.5,    # Precio actual de la acción
    E: 60.0,    # Precio de ejercicio (Strike)
    T: 0.3,     # Tiempo al vencimiento (años)
    r: 0.04,    # Tasa libre de riesgo
    sigma: 0.29,# Volatilidad
    q: 0.0      # Tasa de dividendos (no paga dividendos)
}

d1_expr = (sp.log(S / E) + (r - q + sigma**2 / 2) * T) / (sigma * sp.sqrt(T))
d2_expr = d1_expr - sigma * sp.sqrt(T)

# Calcular los valores numéricos de d1 y d2
d1_val = d1_expr.subs(datos).evalf()
d2_val = d2_expr.subs(datos).evalf()

print(f"d1 = {d1_val}")
print(f"d2 = {d2_val}")

# Calcular N(d1) y N(d2)
def N(x):
    return 0.5 * (1 + sp.erf(x / sp.sqrt(2)))

Nd1_val = N(d1_val).evalf()
Nd2_val = N(d2_val).evalf()

print(f"N(d1) = {Nd1_val}")
print(f"N(d2) = {Nd2_val}")

precio_call_expr = S * sp.exp(-q * T) * N(d1_expr) - E * sp.exp(-r * T) * N(d2_expr)
precio_final = precio_call_expr.subs(datos).evalf()

print(f"El precio de la opción Call es: {precio_final}")
```

### Ecuación Inicial (Black-Scholes)

$$
 \frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + (r - q)S \frac{\partial V}{\partial S} - rV = 0
$$

---

### Paso 1: Transformación del tiempo

Sustitución: \( \tau = T - t \) (por lo tanto, \( dt = -d\tau \))

$$
 -\frac{\partial V}{\partial \tau} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + (r - q)S \frac{\partial V}{\partial S} - rV = 0
$$

$$
 \Rightarrow \frac{\partial V}{\partial \tau} = \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + (r - q)S \frac{\partial V}{\partial S} - rV
$$

---

### Paso 2: Transformación logarítmica

Sustitución: \( S = e^x \) (por lo tanto, \( x = \ln S \)) y se define \( V(S, \tau) = W(x, \tau) \)

$$
 \frac{\partial W}{\partial \tau} = \frac{1}{2}\sigma^2 \frac{\partial^2 W}{\partial x^2} + \left( r - q - \frac{1}{2}\sigma^2 \right)\frac{\partial W}{\partial x} - rW
$$

---

### Paso 3: Transformación exponencial

Sustitución: \( W(x, \tau) = e^{\alpha x + \beta \tau} Z(x, \tau) \)

$$
 \frac{\partial Z}{\partial \tau} = \frac{1}{2}\sigma^2 \frac{\partial^2 Z}{\partial x^2} + \left[ \sigma^2\alpha + r - q - \frac{1}{2}\sigma^2 \right]\frac{\partial Z}{\partial x} + \left[ \frac{1}{2}\sigma^2\alpha^2 + \alpha\left(r - q - \frac{1}{2}\sigma^2\right) - r - \beta \right]Z
$$

Elección de constantes para eliminar los términos de orden inferior:

$$
 \alpha = -\frac{r - q - \frac{1}{2}\sigma^2}{\sigma^2}
$$

$$
 \beta = \frac{1}{2}\sigma^2\alpha^2 + \alpha\left(r - q - \frac{1}{2}\sigma^2\right) - r
$$

---

### Paso 4: Ecuación del calor resultante y Solución

$$
 \frac{\partial Z}{\partial \tau} = \frac{1}{2}\sigma^2 \frac{\partial^2 Z}{\partial x^2}
$$

Solución mediante la fórmula de Green (integral):

$$
 Z(x, \tau) = \frac{1}{\sqrt{2\sigma^2 \pi \tau}} \int_{-\infty}^{\infty} e^{-\frac{(x-s)^2}{2\sigma^2 \tau}} Z(s, 0) \, ds
$$

Al deshacer los cambios (\(Z \rightarrow W \rightarrow V\)):

$$
 V(S, \tau) = \frac{e^{-r\tau}}{\sqrt{2\sigma^2 \pi \tau}} \int_{-\infty}^{\infty} e^{-\frac{(\ln S - s)^2}{2\sigma^2 \tau}} e^{\alpha s} \bar{V}(e^s) \, ds
$$

Fórmula final de Black-Scholes:

$$
 V(S,t) = Se^{-q(T-t)}N(d_1) - Ee^{-r(T-t)}N(d_2)
$$

```python
S_sym, E_sym, T_sym, r_sym, sigma_sym, q_sym = sp.symbols('S E T r sigma q', real=True, positive=True)

d1_expr = (sp.log(S_sym / E_sym) + (r_sym - q_sym + sigma_sym**2 / 2) * T_sym) / (sigma_sym * sp.sqrt(T_sym))
d2_expr = d1_expr - sigma_sym * sp.sqrt(T_sym)

def N(x):
    return 0.5 * (1 + sp.erf(x / sp.sqrt(2)))

call_expr = S_sym * sp.exp(-q_sym * T_sym) * N(d1_expr) - E_sym * sp.exp(-r_sym * T_sym) * N(d2_expr)

call_func = sp.lambdify((S_sym, E_sym, T_sym, r_sym, sigma_sym, q_sym), call_expr, modules=['numpy', 'scipy'])

E_val = 60.0
r_val = 0.04
sigma_val = 0.29
q_val = 0.0


S_array = np.linspace(20, 100, 100)
t_array = np.linspace(0, 0.3, 100)

S_grid, t_grid = np.meshgrid(S_array, t_array)
T_vencimiento = 0.3
T_restante_grid = T_vencimiento - t_grid
Z_grid = call_func(S_grid, E_val, T_restante_grid, r_val, sigma_val, q_val)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(S_grid, t_grid, Z_grid, cmap='viridis', edgecolor='none', alpha=0.9)

fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='Precio de la Opción (V)')

S_actual = 58.5
t_actual = 0.0
T_actual = T_vencimiento - t_actual
V_actual = call_func(S_actual, E_val, T_actual, r_val, sigma_val, q_val)

ax.scatter(S_actual, t_actual, V_actual, color='red', s=50, label=f'Punto Evaluado\nV({S_actual}, 0) = {V_actual:.2f}')

ax.set_xlabel('Precio de la Acción (S)', fontsize=11, labelpad=10)
ax.set_ylabel('Tiempo (t)', fontsize=11, labelpad=10)
ax.set_zlabel('Precio de la Opción (V)', fontsize=11, labelpad=10)
ax.set_title('Superficie de Precios de una Opción Call (Black-Scholes)', fontsize=14)

ax.view_init(elev=25, azim=-60)

ax.legend()
plt.tight_layout()
plt.show()
```
